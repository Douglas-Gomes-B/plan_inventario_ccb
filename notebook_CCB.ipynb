Nesta etapa final, o script padroniza as chaves de ligação para garantir que a associação entre os bens e seus respectivos grupos administrativos seja 100% precisa:
Padronização de Código: Utiliza Regex (r'^BR\s*') para remover o prefixo "BR" e eventuais espaços no início da coluna codco. Isso é necessário porque a base de grupos utiliza apenas o numeral (ex: 11-1234), enquanto a base de dependências contém o prefixo (ex: BR 11-1234).
Integração Final (Left Join): Cruza a tabela mestre (df_merged) com a tabela de setores (df_grupos).
Chave Esquerda: codcoSemBR (Código limpo).
Chave Direita: Codigo CO (ID original do arquivo de grupos).
Resultado: Criação do df_final, que agora possui a hierarquia completa: Bem Imobilizado > Localidade > Setor Regional.

```python
import pandas as pd

# Ler os arquivos Excel
excel_bens = 'C:/Users/dougl/OneDrive/Databricks+CCB+-+2026/Relatório de Bens Imobilizado.xlsx'
excel_dep = 'C:/Users/dougl/OneDrive/Databricks+CCB+-+2026/cadastro_dependencias.xlsx'

df_bens = pd.read_excel(excel_bens)
df_dep = pd.read_excel(excel_dep)

# Transformações necessárias para criar 'codco' em df_dep
# Dividir a coluna 'Casa de Oração' em três colunas temporárias
df_dep[['col1', 'col2', 'col3']] = df_dep['Casa de Oração'].str.extract(r'^(BR \d{2}-\d{4})\s*(-)\s*(.*)$')
# Remover 'col2' e renomear as colunas
df_dep = df_dep.drop(columns=['col2']).rename(columns={'col1': 'codco', 'col3': 'nomeco'})

# Mesclar os DataFrames usando 'Localidade' (df_bens) e 'codco' (df_dep)
df_merged = pd.merge(df_bens, df_dep, left_on='Localidade', right_on='codco', how='left')

display(df_merged.head(10))
```

Este código realiza a leitura e a visualização inicial da terceira base de dados do seu projeto, que contém a estrutura de setorização regional.

```python
excel_grupos = 'C:/Users/dougl/OneDrive/Databricks+CCB+-+2026/grupos_set_lapa.xlsx'
df_grupos = pd.read_excel(excel_grupos)
display(df_grupos.head(10))
```

```python
df_merged['codcoSemBR'] = df_merged['codco'].str.replace(r'^BR\s*', '', regex=True)
display(df_merged.head(10))
```

Este é o comando que consolida toda a sua análise. Ele une a base de Bens + Localidades (que você já tinha no df_merged) com a base de Setores/Grupos.

O que este código faz exatamente:
A Junção (Merge): Ele procura o código da igreja (sem o "BR") na coluna codcoSemBR da sua tabela principal e tenta achar o par correspondente na coluna Codigo CO da tabela de grupos.

A Lógica how='left': Esta é a parte mais segura do código. Ela garante que nenhum item do seu relatório de bens seja apagado, mesmo que aquela igreja específica não conste na planilha de grupos. Se não houver correspondência, as informações do setor aparecerão como NaN (vazio).

A Criação do df_final: Agora você tem um DataFrame completo. Cada linha representa um bem imobilizado, mas agora traz colunas adicionais como o nome do setor, regional ou supervisor (dependendo do que houver no arquivo da Lapa).

```python
df_final = pd.merge(df_merged, df_grupos, left_on='codcoSemBR', right_on='Codigo CO', how='left')
display(df_final.head(10))

```

O que o seu padrão Regex r'^[\w\d\.]+\s*-\s*' faz:
^: Garante que a substituição comece do início da linha.

[\w\d\.]+: Procura por uma sequência de letras, números ou pontos (o \. que você adicionou é ótimo para códigos como 101.001).

\s*-\s*: Localiza o hífen, removendo também qualquer espaço que exista antes ou depois dele.

'' (vazio): Substitui tudo isso por nada, efetivamente "apagando" o código.

```python
df_final['nome_sem_codigo'] = df_final['Nome'].str.replace(r'^[\w\d\.]+\s*-\s*', '', regex=True)
display(df_final.head(20))
```

```python
import pandas as pd
import os
import shutil
import base64
from IPython.display import HTML, display as ipy_display

# --- 1. CONFIGURAÇÃO E LIMPEZA ---
caminho_local = "/tmp/processamento_inventario_final"
arquivo_zip_base = "/tmp/inventario_2026"

if os.path.exists(caminho_local):
    shutil.rmtree(caminho_local)
os.makedirs(caminho_local)

print("1. Gerando arquivos Excel na memória temporária...")

# --- 2. GERAÇÃO DOS ARQUIVOS (PANDAS) ---
count = 0
for nomeco, df_pandas_group in df_selected_dict.items():
    nome_seguro = str(nomeco).replace("/", "-").replace(":", "").strip()
    caminho_arquivo = f"{caminho_local}/Inventario_{nome_seguro}.xlsx"
    
    # df_pandas_group já é um DataFrame do Pandas, não precisa de .toPandas()
    df_pandas_group.to_excel(caminho_arquivo, index=False)
    count += 1

print(f"   -> {count} arquivos gerados.")

# --- 3. COMPACTAÇÃO ---
print("2. Compactando...")
shutil.make_archive(arquivo_zip_base, 'zip', caminho_local)
caminho_zip_final = arquivo_zip_base + ".zip"

# --- 4. PREPARANDO DOWNLOAD DIRETO ---
print("3. Preparando download direto...")
with open(caminho_zip_final, "rb") as f:
    zip_data = f.read()
    b64_zip = base64.b64encode(zip_data).decode()

# --- 5. GERAR BOTÃO ---
html_button = f"""
<div style="text-align:center; margin-top:20px; font-family: sans-serif;">
    <div style="background-color:#e8f5e9; padding:20px; border-radius:10px; border: 1px solid #4caf50;">
        <h3 style="color:#2e7d32;">✅ Pronto! Processamento Concluído Localmente.</h3>
        <p>Arquivo zip gerado com sucesso.</p>
        <br>
        <a href="data:application/zip;base64,{b64_zip}" download="inventario_2026.zip">
            <button style="
                background-color:#2e7d32; 
                color:white; 
                padding:15px 30px; 
                font-size:16px; 
                border-radius:5px; 
                border:none; 
                cursor:pointer; 
                font-weight: bold;
                box-shadow: 0px 4px 6px rgba(0,0,0,0.2);">
                📥 CLIQUE AQUI PARA BAIXAR
            </button>
        </a>
        <p style="color:gray; font-size:12px; margin-top:15px;">Salvar em: D:\\dougl\\Documents\\CCB_Patrimonio\\painelAtivoImob\\inventario_2026</p>
    </div>
</div>
"""

shutil.rmtree(caminho_local)
os.remove(caminho_zip_final)

ipy_display(HTML(html_button))
```
